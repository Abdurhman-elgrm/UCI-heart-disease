```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
Data = pd.read_csv('C:\\Users\\ELGRM\\OneDrive\\Desktop\\UCI heart disease\\data\\raw\\heart_disease_uci.csv')
```

```python
Data.head()
```

```python
Data.drop(columns=['id'], inplace=True)
```

```python
Data.head()
```

```python
Data['chol'].replace(0, np.nan, inplace=True)
Data['trestbps'].replace(0, np.nan, inplace=True)
```

```python
Data.isnull().sum()
```

```python
Data['target'] = Data['num'].apply(lambda x: 1 if x > 0 else 0)
```

```python
Data.head()
```

```python
Data.drop(columns=['num'], inplace=True)
```

```python
Data.head()
```

```python
Data['chol'].fillna(Data['chol'].median(), inplace=True)
Data['trestbps'].fillna(Data['trestbps'].median(), inplace=True)
Data['thalch'].fillna(Data['thalch'].median(), inplace=True)
Data['oldpeak'].fillna(Data['oldpeak'].median(), inplace=True)
```

```python
Data.isnull().sum()
```

```python
categorical_cols = Data.select_dtypes(include='object').columns

for col in categorical_cols:
    Data[col] = Data[col].fillna(Data[col].mode()[0])
```

```python
Data['ca'].fillna(Data['ca'].median(), inplace=True)
```
